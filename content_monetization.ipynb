```python
def channel_size(subs):
    if subs < 10000:
        return 'Nano'        # < 10K
    elif subs < 100000:
        return 'Micro'       # 10K - 100K
    elif subs < 1000000:
        return 'Mid'         # 100K - 1M
    else:
        return 'Mega'        # 1M+

df_mod['channel_size'] = df_mod['subscribers'].apply(channel_size)
```

```python
df_mod
```

```python
def video_length_cat(mins):
    if mins < 5:
        return 'Short'      # Shorts / quick videos
    elif mins < 15:
        return 'Medium'     # Standard videos
    else:
        return 'Long'       # Long form content

df_mod['video_length_cat'] = df_mod['video_length_minutes'].apply(video_length_cat)
```

```python
from scipy import stats
```

```python
#t-test - compares the means of two independent groups to determine if there is a statistically significant difference between them.
#create two independent groups
```

```python
group_weekday = df_mod[df_mod['is_weekend'] == 0]['ad_revenue_usd']
group_weekend = df_mod[df_mod['is_weekend'] == 1]['ad_revenue_usd']
#t_test
tst_stat, p_value = stats.ttest_ind(group_weekday, group_weekend)

print(f"T-test statistic: {tst_stat:.4f}, P-value: {p_value:.4f}") 

if p_value < 0.05:
    print("is_weekend is a useful feature")
else:
    print("is_weekend is not a useful feature")
```

ANNOVA(Analysis of variance)

```python
cat_features.head()
```

```python
cat_features['category'].value_counts()
```

```python
group_education = df_mod[df_mod['category'] == 'Education']['ad_revenue_usd']
group_entertainment = df_mod[df_mod['category'] == 'Entertainment']['ad_revenue_usd']
group_tech = df_mod[df_mod['category'] == 'Tech']['ad_revenue_usd']
#oneway anova test
f_stat, p_value = stats.f_oneway(group_education, group_entertainment, group_tech)
print(f"ANOVA F-statistic: {f_stat:.4f}, P-value: {p_value:.4f}")
if p_value < 0.05:
    print("category is a useful feature")
else:
    print("category is not a useful feature")
```

```python
group_TV = df_mod[df_mod['device'] == 'TV']['ad_revenue_usd']
group_mobile = df_mod[df_mod['device'] == 'Mobile']['ad_revenue_usd']

f_stat, p_value = stats.ttest_ind(group_TV, group_mobile)
print(f"T-test statistic: {f_stat:.4f}, P-value: {p_value:.4f}")

if p_value < 0.05:
    print("device is a useful feature")
else:
    print("device is not a useful feature")
```

```python
group_IN = df_mod[df_mod['country'] == 'IN']['ad_revenue_usd']
group_US = df_mod[df_mod['country'] == 'US']['ad_revenue_usd']

f_stat, p_value = stats.ttest_ind(group_IN, group_US)
print(f"T-test statistic: {f_stat:.4f}, P-value: {p_value:.4f}")
```

```python
group_mid = df_mod[df_mod['channel_size'] == 'Mid']['ad_revenue_usd']
group_mega = df_mod[df_mod['channel_size'] == 'Micro']['ad_revenue_usd']

f_stat, p_value = stats.ttest_ind(group_mid, group_mega)
print(f"T-test statistic: {f_stat:.4f}, P-value: {p_value:.4f}")
```

```python
group_short = df_mod[df_mod['video_length_cat'] == 'Short']['ad_revenue_usd']
group_medium = df_mod[df_mod['video_length_cat'] == 'Medium']['ad_revenue_usd']
group_long = df_mod[df_mod['video_length_cat'] == 'Long']['ad_revenue_usd']
f_stat, p_value = stats.f_oneway(group_short, group_medium, group_long)
print(f"ANOVA F-statistic: {f_stat:.4f}, P-value: {p_value:.4f}")
```

PIPELINE BUILDING

```python
from sklearn.compose import ColumnTransformer  #column transformer is used for merging sub piplines
from sklearn.pipeline import Pipeline #building a pipeline for preprocessing the data
from sklearn.impute import SimpleImputer #fills the null values with mean, median or mode
from sklearn.preprocessing import OneHotEncoder, MinMaxScaler, StandardScaler
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(x_features, y_target, test_size=0.2, random_state=42) #random state maintains same split

# Fit and transform
num_cols = [
    'views', 'likes', 'watch_time_minutes',
     'subscribers','engagement_rate', 'watch_ratio'
]

cat_cols = [
    'category', 'device', 'country'
]

numerical_pipeline = Pipeline(steps=[
    ('scaler', StandardScaler())       # MinMax scaling ✅
])

categorical_pipeline = Pipeline(steps=[
    ('encoder', OneHotEncoder())
])


preprocessor = ColumnTransformer(transformers=[
    ('num', numerical_pipeline, num_cols),
    ('cat', categorical_pipeline, cat_cols)
])

x_train_processed = preprocessor.fit_transform(x_train)
x_test_processed  = preprocessor.fit_transform(x_test)


print(f"X_train shape: {x_train_processed.shape}")
print(f"X_test shape:  {x_test_processed.shape}")
```

```python
from sklearn.linear_model import LinearRegression

# Complete pipeline
full_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('model', LinearRegression())
])

# Train
linear_regression = full_pipeline.fit(x_train, y_train)
linear_regression
```

```python
predict_lr = full_pipeline.predict(x_test)
predict_lr
```

```python
pd.DataFrame({'Actual Revenue': y_test.values, 'Predicted Revenue in LR': predict_lr, 'Error': y_test.values - predict_lr})
```

```python
train_score_lr = full_pipeline.score(x_train, y_train)
test_score_lr = full_pipeline.score(x_test, y_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, mean_absolute_percentage_error
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
```

```python
def full_regression_suite(model, X_train, y_train, X_test, y_test):
    # 1. Prediction Generation
    y_train_pred = model.predict(X_train)
    y_test_pred = model.predict(X_test)
    residuals = y_test.values.ravel() - y_test_pred

    # 2. Metrics Table (Including MAPE)
    data = {
        'MAE': [mean_absolute_error(y_train, y_train_pred), mean_absolute_error(y_test, y_test_pred)],
        'RMSE': [np.sqrt(mean_squared_error(y_train, y_train_pred)), np.sqrt(mean_squared_error(y_test, y_test_pred))],
        'MAPE (%)': [mean_absolute_percentage_error(y_train, y_train_pred)*100, mean_absolute_percentage_error(y_test, y_test_pred)*100],
        'R2': [r2_score(y_train, y_train_pred), r2_score(y_test, y_test_pred)]
    }
    metrics_df = pd.DataFrame(data, index=['Train', 'Test']).T

    print("--- Model Performance ---")
    print(metrics_df.round(3))

    # 3. Visualizations
    fig, axes = plt.subplots(1, 3, figsize=(20, 5))

    # Plot A: Actual vs Predicted
    sns.scatterplot(x=y_test.values.ravel(), y=y_test_pred, alpha=0.4, ax=axes[0])
    axes[0].plot([y_test.min(), y_test.max()], [y_test.min(), y_test.max()], 'r--', lw=2)
    axes[0].set_title('Actual vs. Predicted')

    # Plot B: Residuals
    sns.scatterplot(x=y_test_pred, y=residuals, alpha=0.4, ax=axes[1])
    axes[1].axhline(y=0, color='r', linestyle='--')
    axes[1].set_title('Residuals (Errors) Plot')

    # Plot C: Error Distribution
    sns.histplot(residuals, kde=True, ax=axes[2])
    axes[2].set_title('Distribution of Errors')

    plt.tight_layout()
    plt.show()
```
